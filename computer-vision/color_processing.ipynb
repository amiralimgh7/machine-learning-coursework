## Binarize Image
 Convert the input image to Gray, and then binarize it. do not use built-in functions.

````python
import numpy as np
import cv2
from matplotlib import pyplot as plt




image_manual_binary = np.copy(image)


threshold_value = 127


gray_image = np.mean(image, axis=2).astype(image.dtype)
height, width = gray_image.shape


for i in range(height):
    for j in range(width):
        if gray_image[i, j] > threshold_value:
            image_manual_binary[i, j] = 255
        else:
            image_manual_binary[i, j] = 0

# Display
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(image_manual_binary, cmap='gray')
plt.title('Manually Binarized Image')
plt.axis('off')

plt.show()
````

## Bit-Plane Slicing


Perform Bit-Plane Slicing on the input image.
Implement the function yourself, do not use built-in functions.

````python


height, width = gray_image.shape


bit_planes = np.zeros((8, height, width), dtype=np.uint8)

for bit in range(8):
    for i in range(height):
        for j in range(width):

            bit_planes[bit, i, j] = (gray_image[i, j] >> bit) & 1


bit_planes_images = bit_planes * 255

# Display all bit planes
fig, axes = plt.subplots(2, 4, figsize=(20, 10))
for i, ax in enumerate(axes.flat):
    ax.imshow(bit_planes_images[i, :, :], cmap='gray')
    ax.set_title(f'Bit Plane {i}')
    ax.axis('off')

plt.show()
````

## HSV Color System
In this part we wat to change the RGB Color System to HSV.


````python
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
plt.imshow(hsv)
````

As you can see, cv2.cvtColor() function converts colorspace.
You have to write your own function RGB_to_HSV without using opencv and other built-in functions, to convert an RGB image to an HSV one. compare your results with the above code block.

````python

def BGR_to_RGB(bgr_image):
    rgb_image = bgr_image[:, :, [2, 1, 0]]
    return rgb_image

def RGB_to_HSV(rgb_image):
    hsv_image = np.zeros_like(rgb_image, dtype=np.float32)
    rgb_image = rgb_image.astype('float') / 255.0

    for i in range(rgb_image.shape[0]):
        for j in range(rgb_image.shape[1]):
            r, g, b = rgb_image[i, j]
            max_val = max(r, g, b)
            min_val = min(r, g, b)
            delta = max_val - min_val

            V = max_val
            S = 0 if V == 0 else delta / V

            if delta == 0:
                H = 0
            elif V == r:
                H = (60 * ((g - b) / delta) + 360) % 360
            elif V == g:
                H = (60 * ((b - r) / delta) + 120) % 360
            else:
                H = (60 * ((r - g) / delta) + 240) % 360

            hsv_image[i, j] = [H / 360, S, V]

    return hsv_image
rgb_image = BGR_to_RGB(image)
hsv_image = RGB_to_HSV(rgb_image)
plt.imshow(hsv_image)
plt.title('Custom RGB to HSV Conversion')
plt.axis('off')
plt.show()
````

The HSV color system is used to detect the color as all the chromatic information is in one band.

Here we want to extract the Red color from image.
Research how to do so, write your conclusions and implement it here.


````python

def extract_red(hsv_image):
    mask = np.zeros(hsv_image.shape[:2], dtype=np.uint8)


    lower_red1 = np.array([0 / 360, 0.2, 0.2])
    upper_red1 = np.array([10 / 360, 1.0, 1.0])
    lower_red2 = np.array([170 / 360, 0.2, 0.2])
    upper_red2 = np.array([180 / 360, 1.0, 1.0])

    for i in range(hsv_image.shape[0]):
        for j in range(hsv_image.shape[1]):
            H, S, V = hsv_image[i, j]
            if ((lower_red1[0] <= H <= upper_red1[0] and lower_red1[1] <= S <= upper_red1[1] and lower_red1[2] <= V <= upper_red1[2]) or
                (lower_red2[0] <= H <= upper_red2[0] and lower_red2[1] <= S <= upper_red2[1] and lower_red2[2] <= V <= upper_red2[2])):
                mask[i, j] = 255
    return mask

def apply_mask(rgb_image, mask):
    mask_bool = mask.astype(bool)
    red_parts = np.zeros_like(rgb_image)
    for i in range(3):
        red_parts[:,:,i] = rgb_image[:,:,i] * mask_bool

    return red_parts


hsv_image = RGB_to_HSV(rgb_image)


red_mask = extract_red(hsv_image)


red_parts = apply_mask(rgb_image, red_mask)

# Display
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(rgb_image)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(red_mask, cmap='gray')
plt.title('Red Mask')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(red_parts)
plt.title('Red Parts Isolated')
plt.axis('off')

plt.show()
````

## Histogram Equalization
In this section you will have to perform histogram equalization on the gray image. This transform flattens the gray-level histogram so that all intensities are as equally common as possible. The transform function is a cumulative distribution function (cdf) of the pixel values in the image (normalized to map the range of pixel values to the desired range).

To do so,
1. first display the original histogram of the image using built-in functions,
2. Then write a function to perform Histogram Equalization without using any built-in function, only Numpy.
3. Use the cv2.equalizeHist to compare your results.

````python
gray_image = np.mean(image, axis=2).astype(image.dtype)

def histogram_equalization(image):
    hist, bins = np.histogram(image.flatten(), 256, [0, 256])
    cdf = hist.cumsum()
    cdf_normalized = cdf * hist.max() / cdf.max()  # Normalized (optional)
    cdf_m = np.ma.masked_equal(cdf, 0)
    cdf_m = (cdf_m - cdf_m.min()) * 255 / max(1, cdf_m.max() - cdf_m.min())
    cdf_final = np.ma.filled(cdf_m, 0).astype('uint8')
    equalized_image = cdf_final[image]
    return equalized_image


equalized_image_manual = histogram_equalization(gray_image)


equalized_image_cv = cv2.equalizeHist(gray_image)

fig, axs = plt.subplots(3, 2, figsize=(10, 15))


axs[0, 0].imshow(gray_image, cmap='gray')
axs[0, 0].set_title('Original Image')
axs[0, 0].axis('off')
axs[0, 1].hist(gray_image.ravel(), 256, [0,256])
axs[0, 1].set_title('Original Histogram')

axs[1, 0].imshow(equalized_image_manual, cmap='gray')
axs[1, 0].set_title('Manually Equalized Image')
axs[1, 0].axis('off')
axs[1, 1].hist(equalized_image_manual.ravel(), 256, [0,256])
axs[1, 1].set_title('Manual Equalized Histogram')


axs[2, 0].imshow(equalized_image_cv, cmap='gray')
axs[2, 0].set_title('OpenCV Equalized Image')
axs[2, 0].axis('off')
axs[2, 1].hist(equalized_image_cv.ravel(), 256, [0,256])
axs[2, 1].set_title('OpenCV Equalized Histogram')

plt.tight_layout()
plt.show()
````
